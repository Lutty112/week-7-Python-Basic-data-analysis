```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4))
plt.plot(data.index, data['Age'], label='Age')
plt.title('Cholesterol VS Age Graph')
plt.xlabel('Index (Age)')
plt.ylabel('Cholesterol')
plt.legend()
plt.show()
```

```python
data_grouped = data.groupby('Stage')['Albumin'].mean().reset_index()
plt.bar(data_grouped['Stage'], data_grouped['Albumin'])
plt.title('Cirrhosis Bar Chart') 
plt.xlabel('Stage') 
plt.ylabel('Mean Albumin')
plt.show()
```

```python
plt.hist(data['Platelets'], bins=10, color='lightgreen', edgecolor='black')
plt.title('Histogram of Patients Platelets')
plt.xlabel('Mean')
plt.ylabel('Platelets')
plt.show()
```
